```python
default_balance = 5000
correct_pin = 1234

def withdraw(amount):
    global default_balance
    default_balance -= amount
    return default_balance

def deposit(amount):
    global default_balance
    default_balance += amount
    return default_balance

def check_balance():
    return f"Your balance is {default_balance}"

def verify_pin():
    attempts = 0
    while attempts < 3:
        try:
            entered_pin = int(input("Enter PIN: "))
        except ValueError:
            entered_pin = None          
        if entered_pin == correct_pin:
            return True
        attempts += 1
        print(f"Wrong PIN. {3 - attempts} chance(s) left")
    return False

def start_transaction(option, amount=0):
    if option == 1:
        if amount > default_balance:
            return "Insufficient funds"
        withdraw(amount)
        return f"Transaction completed\nBalance: {default_balance}"
    elif option == 2:                  
        deposit(amount)
        return f"Transaction completed\nBalance: {default_balance}"
    elif option == 3:                  
        return check_balance()
    else:
        return "Something went wrong"
    if verify_pin():
    option = int(input("1.Withdraw  2.Deposit  3.Check balance: "))
    amount = 0
    if option in (1, 2):
        amount = int(input("Enter amount: "))
    print(start_transaction(option, amount))
else:
    print("3 wrong attempts. Transaction stopped")
```
