```python
numbers = [10, 15, 22, 33, 40, 51, 64, 75]
evennumber=[]
oddnumber=[]
for num in numbers:
    if num % 2 == 0:
        evennumber.append(num)
    else:
        oddnumber.append(num)
print("Evennumber:",evennumber)
print("Oddnumber:",oddnumber)
       
```

```python
customers = [
    [1001, "Arun", 25000],
    [1002, "Rahul", 35000],
    [1003, "Bharath", 45000],
    [1004, "Udith", 30000]
]
print("Names:")
for c in customers:
    print(c[1])
salaries = []
for c in customers:
    salaries.append(c[2])
print("Total Salary:", sum(salaries))
print("Highest Salary:", max(salaries))
print("Average Salary:", sum(salaries) / len(salaries))
```

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]
total = 0
largest = matrix[0][0]
for row in matrix:
    for value in row:
        print(value)
        total += value
        if value > largest:
            largest = value

print("Total:", total)
print("Largest:", largest)
```

```python
sales = [
    ("Chennai", 10000),
    ("Mumbai", 15000),
    ("Delhi", 12000),
    ("Bangalore", 18000)
]
total = 0
top_amount = 0
top_city = ""
for city, amount in sales:
    print(f"{city}: {amount}")
    total += amount
    if amount > top_amount:
        top_amount = amount
        top_city = city
print()
print("Total Sales:", total)
print("Highest Sales City:", top_city)
```
